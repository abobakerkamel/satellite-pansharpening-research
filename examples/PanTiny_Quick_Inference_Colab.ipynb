```python
from pathlib import Path
import torch
from models.pantiny.inference import load_model
from models.common.inference_utils import full_scene_tiled_inference
tifs=[n for n in uploaded if n.lower().endswith(('.tif','.tiff'))]
MS_PATH=Path(tifs[0]);PAN_PATH=Path(tifs[1]);OUT=Path('PanTiny_HRMS_6band.tif')
d=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
m=load_model('models/pantiny/weights/best_pantiny_small_6band.pth',d)
full_scene_tiled_inference(m,MS_PATH,PAN_PATH,'weights/train_normalization_stats.json',OUT,tile_lr=160,device=d)
print(OUT)
```
